# 08 — Reusable Code + Paper-Style NOVIC Decoder

## Objectives

The Mini-NOVIC pipeline is now working end-to-end:

- text-only decoder training
- zero-shot transfer to image embeddings
- embedding-noise experiments
- Gaussian noise sweep
- mixed-noise comparison
- greedy decoding
- beam-search decoding

The next stage has two goals.

### Goal 1 — Reusable project code

Stable model, decoding, noise, and utility functions will be moved from
individual notebooks into the `src/` directory of the GitHub repository.

The notebooks will remain responsible for experiments, while `src/` will contain
reusable implementation code.

### Goal 2 — Closer NOVIC architecture

A new decoder will be implemented that more closely follows the WACV NOVIC
paper.

The closer decoder will include:

- bias-free prefix projection
- learned positional embeddings
- pre-LayerNorm Transformer blocks
- GELU activation
- dropout
- full attention between prefix tokens
- causal attention for generated noun tokens
- tied input token embeddings and output projection weights
- bias-free linear layers
- bias-free LayerNorm
- contractive feed-forward layers
- shifted autoregressive prediction without a BOS token

The existing Mini-NOVIC decoder will be retained as a baseline.

This notebook first verifies the reusable project structure, then implements and
tests the new paper-style decoder.
